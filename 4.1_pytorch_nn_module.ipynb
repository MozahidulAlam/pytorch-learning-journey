```python
# create model class
import torch
import torch.nn as nn

# inharit from nn.Module class
class Model(nn.Module):

  def __init__(self, num_features):

    super().__init__()

    self.linear = nn.Linear(num_features, 1)
    self.sigmoid = nn.Sigmoid()


  def forward(self, features):

    out = self.linear(features)
    out = self.sigmoid(out)

    return out
```

```python
# create dataset
features = torch.rand(10,5)

# create model
model = Model(features.shape[1])

# call model for forward pass
# model.forward(features)
model(features)
```

```python
# create model class
import torch
import torch.nn as nn

# inharit from nn.Module class
class Model(nn.Module):

  def __init__(self, num_feature):

    super().__init__()

    self.linear1 = nn.Linear(num_feature, 3)#hidden layer input 5,output3
    self.relu = nn.ReLU() #activitaion function on hidden layer relu
    self.linear2 = nn.Linear(3,1)#input 3 and output 1
    self.sigmoid = nn.Sigmoid()#output between 0 and 1

  def forward(self, feature):

    out = self.linear1(feature)
    out = self.relu(out)
    out = self.linear2(out)
    out = self.sigmoid(out)

    return out
```

```python
# create dataset
feature = torch.rand(10,5)

# create model
model = Model(feature.shape[1])

# call model for forward pass
# model.forward(features)
model(feature)
```

```python
# create model class
import torch
import torch.nn as nn

class Model(nn.Module):

  def __init__(self, num_features):

    super().__init__()
    self.network = nn.Sequential(
        nn.Linear(num_features, 3),
        nn.ReLU(),
        nn.Linear(3, 1),
        nn.Sigmoid()
    )

  def forward(self, features):

    out = self.network(features)

    return out
```

```python
# create dataset
features = torch.rand(10,5)

# create model
model = Model(features.shape[1])

# call model for forward pass
# model.forward(features)
model(features)
```
