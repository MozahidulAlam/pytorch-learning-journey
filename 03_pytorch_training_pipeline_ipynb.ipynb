## Weights and Bias in Neural Networks

**Weights:** Each input feature has a corresponding weight. Weights determine how much each feature contributes to the prediction.

**Bias:** An additional learnable value added to the weighted sum to adjust the model's output.

**Formula:**
`z = (X * W) + b`

**Why random weights?** Weights start with random values so neurons can learn different patterns during training.

**Why zero bias?** Zero is a common, simple starting value for bias. Training can adjust it later.

**During training:**
1. Calculate prediction.
2. Calculate loss.
3. Use gradients to update weights and bias.
4. Repeat to learn better predictions.

**Remember:**
- Weights = Feature contribution
- Bias = Output adjustment
- Loss = Prediction error measure
- Gradients = Help update parameters

#Defining the model

```python
class MySimpleNN():

  def __init__(self, X):

    self.weights = torch.rand(
        X.shape[1], 1,
        dtype=torch.float64,
        requires_grad=True
        )
    self.bias = torch.zeros(
        1,
        dtype=torch.float64,
        requires_grad=True
        )

  def forward(self, X):
    z = torch.matmul(X, self.weights) + self.bias
    y_pred = torch.sigmoid(z)
    return y_pred

  def loss_function(self, y_pred, y):
    # Clamp predictions to avoid log(0)
    epsilon = 1e-7
    y_pred = torch.clamp(y_pred, epsilon, 1 - epsilon)

    # Calculate loss
    loss = -(y_train_tensor * torch.log(y_pred) + (1 - y_train_tensor) * torch.log(1 - y_pred)).mean()
    return loss



```

#Important Parameters

```python
learning_rate = 0.1
epochs = 25
```

#Training Pipeline

```python
# create model
model = MySimpleNN(X_train_tensor)

# define loop
for epoch in range(epochs):

  # forward pass
  y_pred = model.forward(X_train_tensor)

  # loss calculate
  loss = model.loss_function(y_pred, y_train_tensor)

  # backward pass,all derivatives calculate
  loss.backward()

  # parameters update
  with torch.no_grad():
    model.weights -= learning_rate * model.weights.grad
    model.bias -= learning_rate * model.bias.grad

  # zero gradients
  model.weights.grad.zero_()
  model.bias.grad.zero_()

  # print loss in each epoch
  print(f'Epoch: {epoch + 1}, Loss: {loss.item()}')
```

#Evaluation

```python
# model evaluation
with torch.no_grad():
  y_pred = model.forward(X_test_tensor)
  y_pred = (y_pred > 0.9).float()
  accuracy = (y_pred == y_test_tensor).float().mean()
  print(f'Accuracy: {accuracy.item()}')
```

Input X — patient-এর feature values।



Weights W — কোন feature prediction-এ কতটা প্রভাব ফেলবে, তা শেখে।



Bias b — weighted result-কে adjust করতে শেখে।



Sigmoid — result-কে 0 এবং 1-এর মধ্যে নিয়ে আসে।



Loss — prediction আর actual answer-এর মধ্যে কতটা ভুল, তা মাপে।



Backpropagation — ভুল কমানোর জন্য weights ও bias কীভাবে বদলানো উচিত, তার gradients হিসাব করে।



Parameter update — gradients ব্যবহার করে weights ও bias পরিবর্তন করে।



Epoch — training loop-এর একটি iteration; এখানে পুরো training dataset একবার ব্যবহার করা হয়।
